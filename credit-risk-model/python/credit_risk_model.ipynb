```python
# Step 7 — Train test split

from sklearn.model_selection import train_test_split

X = df_model.drop("loan_status", axis=1)
y = df_model["loan_status"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)
```

```python
# clean training data
X_train = X_train.fillna(0)

# clean test data
X_test = X_test.fillna(0)
```

```python

print(X_test.isnull().sum())
```

```python
# Step 8 — Train Logistic Regression model

from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=500)

model.fit(X_train, y_train)
```

```python
# Step 9 — Make predictions

y_pred = model.predict(X_test)

y_prob = model.predict_proba(X_test)[:,1]
```

```python
# Step 10 — Evaluate model

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
# Confusion matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
# Classification report

print(classification_report(y_test, y_pred))
```

```python
# Step 11 — Feature importance

importance = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_[0]
})

importance.sort_values(by="Coefficient", ascending=False)
```

```python
# Step 12 — Predict default probability for customers

results = X_test.copy()

results["actual"] = y_test

results["predicted"] = y_pred

results["default_probability"] = y_prob

results.head()
```

```python
# Step 13 — Business insight example -Example interpretation:

# Customers with:

# high loan amount
# high interest rate
# low income

# have higher probability of default.
```
